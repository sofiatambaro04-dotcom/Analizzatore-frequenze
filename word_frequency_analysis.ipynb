```python
import nltk
import re
from collections import Counter

nltk.download("gutenberg")
from nltk.corpus import gutenberg

testo = gutenberg.raw("austen-emma.txt").lower()

parole = re.findall(r"[a-z']+", testo)
frequenze = Counter(parole)

print("Parole totali:", len(parole))
print("Parole diverse:", len(frequenze))
print("Type/token ratio:", round(len(frequenze) / len(parole), 3))

for parola, n in frequenze.most_common(20):
    print(parola, n)
```

```python
nltk.download("stopwords")
from nltk.corpus import stopwords

stop = set(stopwords.words("english"))

parole_utili = [p for p in parole if p not in stop and len(p) > 2]
frequenze_utili = Counter(parole_utili)

print("Parole utili:", len(parole_utili))
for parola, n in frequenze_utili.most_common(20):
    print(parola, n)
```
